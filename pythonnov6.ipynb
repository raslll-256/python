```python
a = list(map(int, input("Enter first list: ").split(",")))
b = list(map(int, input("Enter second list: ").split(",")))

if len(a) == len(b):
    print("Lists are same length")
else:
    print("Lists are different length")

if sum(a) == sum(b):
    print("Lists having same sum")
else:
    print("Lists having different sum")

for i in a:
    if i in b:
        print("Common value: ",i)
```
